```python
import pandas as pd

df = pd.read_csv("data/legend.csv")
df.head()
```

```python
def hex_to_rgb(hex_color):
    hex_color = hex_color.lstrip('#')
    return tuple(int(hex_color[i:i+2], 16) for i in (0, 2, 4))

df["rgb"] = df["hex"].apply(hex_to_rgb)
target_colors = df["rgb"].tolist()
target_colors[:3]
```

```python
import numpy as np
import cv2
from tqdm import tqdm

img = cv2.imread("data/el_harrach_highres_map.png")
img_i16 = img.astype(np.int16)

colors = np.array(target_colors, dtype=np.int16)[:, ::-1]  # RGB -> BGR
tolerance = 0

overlays = []

for i, c in enumerate(tqdm(colors, desc="Generating overlays")):
    mask = np.all(np.abs(img_i16 - c) <= tolerance, axis=2)

    result = img.copy()
    result[mask] = (0, 0, 255)

    overlays.append(result)

    rgb = c[::-1].astype(int)
    hex_color = "#{:02X}{:02X}{:02X}".format(rgb[0], rgb[1], rgb[2])
    hex_id = hex_color.replace("#", "")

    cv2.imwrite(f"output/overlay_{i+1}_{hex_id}.png", result)
```
